功能,命令
新建会话,tmux new -s <会话名> (建议给训练任务起名，如 train)
列出所有会话,tmux ls
进入/恢复会话,tmux attach -t <会话名> (简写 tmux a -t ...)
重命名会话,tmux rename-session -t <旧名> <新名>
关闭/删除会话,tmux kill-session -t <会话名>
